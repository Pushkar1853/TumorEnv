```python
import pandas as pd
import torch
```

```python
def initialize_location_matrix(shape, params):
    file_path = params['file_path']

    if file_path[-3:] == 'csv':
        data = pd.read_csv(file_path)
    
    data_values = data.values
    # tumor density map -> (height, width)
    # sum of all ones in the map as the number of agents

    grid_height, grid_width = shape
    num_agents = (data_values).sum()
    data_values_tensor = torch.tensor(data_values, dtype=torch.float32)
    var = torch.reshape(data_values_tensor, (grid_height, grid_width))

    print("num_agents: ", num_agents)
    location_matrix = torch.zeros(num_agents, grid_height, grid_width)

    agent_idx = 0
    for i in range(grid_height):
        for j in range(grid_width):
            if var[i][j] == 1:
                location_matrix[agent_idx][i][j] = 1
                agent_idx += 1

    return location_matrix
```

```python
location_matrix = initialize_location_matrix((100, 100), {'file_path': file_path})
```
